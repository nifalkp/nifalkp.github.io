```python

innings_list = []

for (match_id, inn), grp in samson.groupby(['ID','Innings']):
    runs      = grp['BatsmanRun'].sum()
    balls     = len(grp)
    dismissed = int(grp['IsWicketDelivery'].sum() > 0)
    innings_list.append({
        'match_id' : match_id,
        'innings'  : inn,
        'runs'     : runs,
        'balls'    : balls,
        'dismissed': dismissed
    })

innings_df = pd.DataFrame(innings_list)
print(innings_df.head(10))
```

```python
ducks    = (innings_df['runs'] == 0).sum()
fifties  = ((innings_df['runs'] >= 50) & (innings_df['runs'] < 100)).sum()
hundreds = (innings_df['runs'] >= 100).sum()
highest  = innings_df['runs'].max()

print("===== MILESTONE SCORES =====")
print("Ducks    :", ducks)
print("50s      :", fifties)
print("100s     :", hundreds)
print("Highest  :", highest)
```

```python
merged= ball.merge(match,left_on='ID',right_on='match_number',how='left')
samson_full = merged[merged['Batter'] =='SV Samson'].copy()
```

```python
print(samson_full.shape)
```

```python
print(samson_full[['ID','match_date','BatsmanRun']].head())
```

```python
samson_full['season']=pd.to_datetime(samson_full['match_date']).dt.year
print(samson_full['season'].head())
```

```python
season_stats = samson_full.groupby('season').agg(
    balls=('BatsmanRun','count'),
    runs=('BatsmanRun','sum'),
    dismissed =('IsWicketDelivery','sum')
).reset_index()
print(season_stats)
```

```python
season_stats['strike_rate'] = (season_stats['runs']/season_stats['balls']*100).round(2)
season_stats['average']=(season_stats['runs']/season_stats['dismissed']).round(2)
print(season_stats)

```

```python
samson_full['ball_seq'] = samson_full['Overs'] * 6 + samson_full['BallNumber']
ball['ball_seq'] = ball['Overs'] * 6 + ball['BallNumber']

first_ball = samson_full.groupby(['ID','Innings'])['ball_seq'].min().reset_index()
print(first_ball.head())
```

```python
def batting_position(match_id, innings, first_seq):
    before_samson = ball[
        (ball['ID'] == match_id) &
        (ball['Innings'] == innings) &
        (ball['ball_seq'] < first_seq)
    ]
    batters_before = before_samson['Batter'].nunique()
    return batters_before + 1

first_ball['position'] = first_ball.apply(
    lambda row: batting_position(row['ID'], row['Innings'], row['ball_seq']),
    axis=1
)

print(first_ball.head(10))
```

```python
first_ball = first_ball.merge(
    samson_full[['ID','Innings','season']].drop_duplicates(),
    on=['ID','Innings'],
    how='left'
)
print(first_ball.head(10))
```

```python
position_by_season = first_ball.groupby('season')['position'].agg(
    lambda x: x.value_counts().idxmax()
).reset_index()

position_by_season.columns = ['season', 'most_common_position']
print(position_by_season)
```

```python
def get_phase(over):
    if over < 6:
        return 'Powerplay'
    elif over <15:
        return 'Middle'
    else:
        return'Death'
```

```python
samson=samson.copy()
```

```python
samson['phase'] = samson['Overs'].apply(get_phase)
```

```python
print (samson['phase'].value_counts())
```

```python
phase_stats = samson.groupby('phase').agg(
    balls = ('BatsmanRun', 'count'),
    runs  = ('BatsmanRun', 'sum'),
    fours = ('BatsmanRun', lambda x: (x==4).sum()),
    sixes = ('BatsmanRun', lambda x: (x==6).sum()),
    dots  = ('BatsmanRun', lambda x: (x==0).sum())
).reset_index()
```

```python
phase_stats['strike_rate']=(phase_stats['runs']/phase_stats['balls']*100).round(2)
```

```python
# ===== MILESTONES - 50s, 100s, Highest, Ducks per season =====
season_50s = []

for season, grp in samson_full.groupby('season'):
    innings_list = []
    for (match_id, inn), inn_grp in grp.groupby(['ID','Innings']):
        runs = inn_grp['BatsmanRun'].sum()
        innings_list.append(runs)
    
    innings_runs = pd.Series(innings_list)
    season_50s.append({
        'season'  : season,
        'fifties' : ((innings_runs >= 50) & (innings_runs < 100)).sum(),
        'hundreds': (innings_runs >= 100).sum(),
        'highest' : innings_runs.max(),
        'ducks'   : (innings_runs == 0).sum()
    })

milestones_df = pd.DataFrame(season_50s)
print('===== MILESTONES PER SEASON =====')
print(milestones_df)
milestones_df.to_csv('samson_milestones.csv', index=False)

# ===== OVERALL MILESTONES =====
all_innings = []
for (match_id, inn), inn_grp in samson_full.groupby(['ID','Innings']):
    runs = inn_grp['BatsmanRun'].sum()
    all_innings.append(runs)

all_innings_runs = pd.Series(all_innings)
print()
print('===== OVERALL CAREER MILESTONES =====')
print('Fifties  :', ((all_innings_runs >= 50) & (all_innings_runs < 100)).sum())
print('Hundreds :', (all_innings_runs >= 100).sum())
print('Highest  :', all_innings_runs.max())
print('Ducks    :', (all_innings_runs == 0).sum())

# ===== WICKETKEEPING - Catches and Stumpings =====
samson_fielding = ball[ball['FieldersInvolved'].str.contains('Samson', na=False)].copy()

fielding_stats = samson_fielding['Kind'].value_counts().reset_index()
fielding_stats.columns = ['dismissal_type', 'count']

print()
print('===== WICKETKEEPING STATS =====')
print(fielding_stats)

fielding_stats.to_csv('samson_fielding.csv', index=False)

print()
print('All files exported!')
```
